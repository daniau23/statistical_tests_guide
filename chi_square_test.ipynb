```python
import numpy as np
from scipy import stats
```

```python
# 1. Define the observed contingency table as a 2D array or list
# Rows: Gender (Male, Female)
# Columns: Products (A, B, C)'
observed = np.array([
    [50,30,20],
    [30,40,30],
])


# Run the chi_square test of independence
chi2_stat, p_value, dof, expected = stats.chi2_contingency(observed)
```

```python
print(f"Chi_Square Statistic: {round(chi2_stat,2)}")
print(f"Degrees of Freedom: {dof}")
print(f"p_value: {round(p_value,2)}")
```

```python
alpha = 0.05
if p_value < alpha:
    print(f"\nResult: Reject Null hypothesis (p-value {round(p_value,2)} < {alpha}).")
    print("Conclusion: Product preference is significantly associated with gender.")
else:
    print(f"\nResult: Fail to reject Null hypothesis (p-value {round(p_value,2)} >= {alpha}).")
```
