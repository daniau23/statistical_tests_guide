```python
import numpy as np
import scipy.stats as stats
```

```python
# Suppose we have a new and old same and
# we what to evaluate the average time 
# for a task to be completed in minutes
control = np.array([120, 130, 125, 140, 135])
treatment = np.array([105, 110, 100, 115, 105])
```

```python
# Independent Two-Sample T-Test (Equal Variance / Student's t-test)
# alternative='less' tests if Treatment < Control (H1: mu_T < mu_C)
t_stat, p_value = stats.ttest_ind(treatment, 
                            control, 
                            equal_var=True, # Matches manual pooled variance calculation
                            alternative='less') # One-sided test (Treatment faster than Control)
```

```python
print("--- Student's T-Test (Equal Variances Assumed) ---")
print(f"Treatment Mean : {treatment.mean():.2f} sec")
print(f"Control Mean   : {control.mean():.2f} sec")
print(f"t-statistic    : {round(t_stat, 2)}")
print(f"p-value        : {round(p_value, 2)}")
```

```python
alpha = 0.05
if p_value < alpha:
    print("\nResult: Reject Null Hypothesis (p < 0.05)")
    print("Conclusion: Significant evidence that the new system reduces task time.")
else:
    print("\nResult: Fail to reject Null Hypothesis.")
```

### **For a Larger Dataset in production**   

We pass equal_var=False to run Welch's t-test to protect against unequal group variances

```python
# Independent Two-Sample T-Test (Equal Variance / Student's t-test)
# alternative='less' tests if Treatment < Control (H1: mu_T < mu_C)
# Welch's T-Test (Best practice for production A/B tests)
t_stat, p_value = stats.ttest_ind(treatment, 
                            control, 
                            equal_var=False, # Matches Welch's t-test
                            alternative='less') # One-sided test (Treatment faster than Control)
```

```python
print("--- Student's T-Test (Equal Variances Assumed) ---")
print(f"Treatment Mean : {treatment.mean():.2f} sec")
print(f"Control Mean   : {control.mean():.2f} sec")
print(f"t-statistic    : {round(t_stat, 2)}")
print(f"p-value        : {round(p_value, 2)}")
```

```python
alpha = 0.05
if p_value < alpha:
    print("\nResult: Reject Null Hypothesis (p < 0.05)")
    print("Conclusion: Significant evidence that the new system reduces task time.")
else:
    print("\nResult: Fail to reject Null Hypothesis.")
```
