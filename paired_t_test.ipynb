```python
import numpy as np
import scipy.stats as stats
```

```python
# Suppose we have these from the validation set
# Because test set should be unseen data, we will use the validation set to simulate the test set for this example.
# 1. Define actual demand (y_test) and model predictions on the same test observations
y__val_test = np.array([650.0, 545.0, 780.0, 610.0, 715.0, 605.0, 655.0, 565.0, 730.0, 810.0])

predictions = {
    "Random Forest": np.array([639.7, 559.6, 761.8, 601.6, 701.2, 596.6, 644.7, 573.4, 718.3, 791.8]),
    "Linear Regression": np.array([668.2, 523.5, 815.4, 597.7, 735.4, 582.8, 672.2, 541.5, 755.2, 845.4])
}
```

```python
best_model = "Random Forest" # Based on the r2 and RMSE
competitor = "Linear Regression"
```

```python
# Absolute errors = |Actual - Predicted|
absolute_errors = {} # For compiling the errors
for model_name, preds in predictions.items():
    absolute_errors[model_name] = np.abs(y__val_test - preds)
```

```python
# Extract error vectors for the paired test
best_errors = absolute_errors[best_model]
competitor_errors = absolute_errors[competitor]
```

```python
# Perform paired t-test
t_stat, p_value = stats.ttest_rel(competitor_errors, best_errors)
```

```python
# Interpret the results
alpha = 0.05
decision = "Reject Null hypothesis" if p_value < alpha else "Fail to reject Null hypothesis"
```

```python
print(f"--- Paired t-test: {best_model} vs {competitor} ---")
print(f"Absolute errors ({best_model})     : {np.round(best_errors, 2)}")
print(f"Absolute errors ({competitor}): {np.round(competitor_errors, 2)}")
print(f"\nResults:")
print(f"  • t-statistic : {round(t_stat,2)}")
print(f"  • p-value     : {round(p_value,2)}")
print(f"  • Decision    : {decision}")
```

```python
if p_value < alpha:
    print("Conclusion: Statistically significant evidence of a difference in predictive accuracy.")
else:
    print("Conclusion: Insufficient statistical evidence of a difference in predictive accuracy.")
```
