```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest
```

```python
# Order: [Treatment (Fast delivery), Control (Standard delivery)]
successes = np.array([4200,4800]) # Churned users in each group
nobs = np.array([10000,10000]) # Total users in each group
```

```python
z_stat, p_value  = proportions_ztest(successes, nobs,alternative="smaller")
```

```python
print(f"Z-statistic: {round(z_stat,2)}")
print(f"p-value: {round(p_value,2)}")
```

```python
alpha = 0.05
if p_value < alpha:
    print("Reject Null Hypothesis: Faster delivery significantly reduces churn")
else:
    print("Fail to reject Null Hypothese")
```
